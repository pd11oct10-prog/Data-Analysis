Function Application and Mapping

```python
import pandas as pd
import numpy as np
from pandas import Series,DataFrame

frame = pd.DataFrame(np.random.randn(4,3),
                     columns=list("bde"),
                     index=["Utah", "Ohio", "Texas", "Oregon"])
frame
np.abs(frame)

def f1(x):
    return x.max() - x.min()

frame.apply(f1) #NOTE: This is with respect to the original frame not the positive np.abs(frame)
frame.apply(f1,axis="columns")

def f2(x):
    return pd.Series([x.max(),x.min()],index=["min","max"])

frame.apply(f2)

def my_format(x):
    return f"{x:.2f}"

frame.map(my_format)

frame["e"].map(my_format)
```

Sorting and Ranking

```python
import pandas as pd
import numpy as np
from pandas import Series,DataFrame

#SORTING

obj = pd.Series(np.arange(4), index=["d", "a", "b", "c"])
obj

obj.sort_index()

frame = pd.DataFrame(np.arange(8).reshape((2, 4)),
            index=["three", "one"],
            columns=["d", "a", "b", "c"])

frame.sort_index()
frame = frame.sort_index(axis="columns",ascending=True)
frame

frame = frame.reindex(["one","three"])
frame

frame = frame.sort_index(axis="columns",ascending=False)
frame

obj = pd.Series([4, 7, -3, 2])
obj

obj.sort_values()

obj = pd.Series([4, np.nan, 7, np.nan, -3, 2])
obj.sort_values()

obj.sort_values(na_position="first")

frame = pd.DataFrame({"b" : [4,7,-3,2],"a":[41,5,5,2]})
frame
frame.sort_values("b")
frame.sort_values #this won't do shit try this

frame.sort_values(["b","a"]) #but it might sort the first column written here it is b


#RANKING

obj = pd.Series([7, -5, 7, 4, 2, 0, 4])
obj.rank()

obj.rank(method="first")
obj.rank(method="max")

frame = pd.DataFrame({"b": [4.3, 7, -3, 2], "a": [0, 1, 0, 1],
         "c": [-2, 5, 8, -2.5]})
frame

frame.rank(axis="columns")
frame.rank(axis="index")

```
