### 4. Ticks, Labels, and Legends ###

```python
fig, ax = plt.subplots()
ax.plot(np.random.randn(1000).cumsum())
ax.set_title("Parabola")
ax.set_xlabel("Input")
ax.set_ylabel("Output")
ax.set_xticks([0, 250, 500, 750, 1000])
ax.set_xticklabels(["one", "two", "three", "four", "five"])
fig , axe = plt.subplots()
axe.plot(np.random.randn(1000).cumsum())
axe.set(title="Second Plot",xlabel="Stages",ylabel="Values",xticks=[0, 250, 500, 750, 1000],xticklabels=["one", "two", "three", "four", "five"])
fig.show()
```

Legends

```python
fig, ax = plt.subplots()
ax.plot(np.random.randn(1000).cumsum(), color="black", label="one")
ax.plot(np.random.randn(1000).cumsum(), color="blue", linestyle="dashed", label="two")
ax.legend(loc="best")
```

```python
tips = pd.read_csv("examples/tips.csv")
tips.head()
```

```python
party_counts = pd.crosstab(tips["day"], tips["size"])
party_counts
```

```python
sorted_days = pd.Index(["Thur","Fri","Sat","Sun"])
party_counts = party_counts.reindex(sorted_days)
party_counts = party_counts.loc[:,2:5]
party_counts
```

```python
#Normalize
party_pcts = party_counts.div(party_counts.sum(axis=1),
                              axis=0)
party_pcts
```

```python
plt.style.use("grayscale")
party_pcts.plot.bar(stacked=True).set_ylim(0,1.2)
```

Scatter or Point Plots

```python
macro = pd.read_csv("examples/macrodata.csv")
macro.head()
```

```python
data = macro[["cpi", "m1", "tbilrate", "unemp"]]
data.head()
```

```python
trans_data = np.log(data).diff().dropna()
trans_data.tail()
```

```python
ax = sns.regplot(x="m1",y="unemp",data=trans_data)
ax.set_title("Changes in log(m1) versus log(unemp)")
```

```python
sns.pairplot(trans_data,diag_kind="kde",plot_kws={"alpha":0.7}) #NOTE: CHECK OUT MORE ABOUT THIS WHEN YOU NEED IT THIS SHIT"S IMPORTANT I THINK
```

Facet grids and categorical data

```python
sns.catplot(x="day",y="tip_pct",hue="time",col="smoker",
            kind="bar",data=tips[tips["tip_pct"] < 1])
```

```python
sns.catplot(x="day",y="tip_pct",row="time",col="smoker",
            kind="bar",data=tips[tips["tip_pct"] < 1])
```
