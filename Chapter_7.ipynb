Handling Missing Data

```python
import pandas as pd
from pandas import DataFrame,Series
import numpy as np

float_data = pd.Series([1.2, -3.5, np.nan, 0])
float_data

float_data.isna()

string_data = pd.Series(["aardvark", np.nan, None, "avocado"])
string_data

string_data.isna()

data1 = pd.Series([1, np.nan, 3.5, np.nan, 7])
data1

data1.dropna()
data1[data1.notna()] #same as above

#CONTINUE FROM 224

data = pd.DataFrame([[1., 6.5, 3.], [1., np.nan, np.nan],
                     [np.nan, np.nan, np.nan], [np.nan, 6.5, 3.]])
data

data.dropna()
data.dropna(how="all")

data[4] = np.nan #this will add a new COLUMN with NaN values
# data.loc[4] = np.nan #this will add a new ROW with NaN values
data

data = data.dropna(how="all",axis="columns")
print(data)

df = pd.DataFrame(np.random.randn(7,3))
print(df)

df.iloc[:4,1] = np.nan
df.iloc[:2,2] = np.nan
print(df)

df.dropna()

df.dropna(thresh=0) #The thresh parameter does not count how many NaN values to remove. Instead, it specifies the 
                    #minimum number of valid, non-missing values a row must have to be kept. if i set thresh = 1 then there must be atleast one non-NaN value row

df.fillna(0)
df.fillna({1: 0.5, 2: 0})
df.ffill()

data1 = pd.Series([1., np.nan, 3.5, np.nan, 7])
data1.fillna(data1.mean())
data[data.notna()]
```

Regular Expressions

```python
import re

text = "foo bar\t baz \tqux"
text

re.split(r"\s+",text)

regex = re.compile(r"\s+")
regex.split(text)
regex.findall(text)

text = """Dave dave@google.com
Steve steve@gmail.com
Rob rob@gmail.com
Ryan ryan@yahoo.com"""

pattern = r"[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,4}"

regex = re.compile(pattern,flags=re.IGNORECASE)
regex.findall(text)

m = regex.search(text)
m

text[m.start():m.end()] #note this will look at Dave dave@google.com and if you write m.start() which is at 5th position 
                        #and m.end() is where the index ends which is 20 (right after com) so writing [m.start():6] will give you letter 'd'

print(regex.match(text))
print(regex.sub("SAME", text))

pattern = r"([A-Z0-9._%+-]+)@([A-Z0-9.-]+)\.([A-Z]{2,4})" #NOTE : THIS IS DIFFERENT FROM THE AABOE
regex = re.compile(pattern, flags=re.IGNORECASE)

m = regex.match("wesm@bright.net")
m
m.groups()

regex.findall(text)

print(regex.sub(r"Username: \1, Domain: \2, Suffix: \3", text)) # HERE SPECIAL SYMBOLS \1 MATCH THE GROUPS FIRST PART AND SO ON
```

Computations with Categoricals

```python
rng = np.random.default_rng(seed=12345)
draws = rng.standard_normal(1000)
draws[:5]

bins = pd.qcut(draws,4)
bins

bins = pd.qcut(draws, 4, labels=['Q1', 'Q2', 'Q3', 'Q4'])
bins

bins.codes[:10]

bins = pd.Series(bins,name='quartile')

results = pd.Series(draws).groupby(bins).agg(['count','min','max']).reset_index()
results
results['quartile']
```

Better performance with categoricals

```python
N = 10_000_000
labels = pd.Series(['foo', 'bar', 'baz', 'qux'] * (N // 4))

categories = labels.astype('category')

labels.memory_usage(deep=True)
categories.memory_usage(deep=True)

%time _  = labels.astype('category')
%time _

%timeit labels.value_counts()
%timeit categories.value_counts()
```
